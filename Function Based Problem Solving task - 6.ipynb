Problem Statement 1:Smart Parking Lot Management System

```python
def regulate_smart_parking(movement_sequence, parking_limit):
    live_vehicle_tally = 0
    highest_occupancy_recorded = 0
    for vehicle_event in movement_sequence:
        if vehicle_event == "IN":
            live_vehicle_tally += 1
            if live_vehicle_tally > highest_occupancy_recorded:
                highest_occupancy_recorded = live_vehicle_tally
        elif vehicle_event == "OUT":
            live_vehicle_tally -= 1
    print("Currently Parked Vehicles:", live_vehicle_tally)
    if live_vehicle_tally > parking_limit:
        print("Parking Status: Capacity Exceeded")
    else:
        print("Parking Status: Available")
    print("Peak Parking Usage:", highest_occupancy_recorded)

maximum_slot_capacity = 50
entry_exit_stream = ["IN", "IN", "IN", "OUT", "IN", "IN", "OUT"]

regulate_smart_parking(entry_exit_stream, maximum_slot_capacity)
```

 Problem Statement 2:Online Food Delivery Time Estimator

```python
def generate_food_eta(distance_measure, traffic_label, weather_label):
    base_duration = distance_measure * 5
    traffic_multiplier = 1
    weather_multiplier = 1
    if traffic_label == "High":
        traffic_multiplier = 1.25
    elif traffic_label == "Medium":
        traffic_multiplier = 1.10
    if weather_label == "Rainy":
        weather_multiplier = 1.20
    elif weather_label == "Stormy":
        weather_multiplier = 1.40
    adjusted_time = base_duration * traffic_multiplier * weather_multiplier
    return round(adjusted_time)
delivery_distance_km = 8
traffic_level_entry = "High"
weather_type_entry = "Rainy"
final_eta_minutes = generate_food_eta(delivery_distance_km, traffic_level_entry, weather_type_entry)
print("Estimated Delivery Time:", final_eta_minutes, "minutes")
```

Problem Statement 3:Movie Theatre Seat Occupancy

```python
def evaluate_show_occupancy(total_seat_capacity, booking_data_list):
    booked_seat_count = 0
    for booking in booking_data_list:
        if booking == 1:
            booked_seat_count += 1
    occupancy_percentage_value = (booked_seat_count / total_seat_capacity) * 100
    print("Occupancy:", round(occupancy_percentage_value), "%")
    if occupancy_percentage_value == 100:
        print("Show Status: Housefull")
    elif occupancy_percentage_value >= 75:
        print("Show Status: Almost Full")
    else:
        print("Show Status: Seats Available")
    if occupancy_percentage_value >= 90:
        print("Suggestion: Consider Opening Additional Show")
total_theatre_seats = 200
seat_booking_records = [1] * 150

evaluate_show_occupancy(total_theatre_seats, seat_booking_records)
```

Problem Statement 4:Cloud Server Load Classification System

```python
def classify_server_load(cpu_readings_list):
    total_cpu_usage_value = 0
    for usage_value in cpu_readings_list:
        total_cpu_usage_value += usage_value
    number_of_readings_count = len(cpu_readings_list)
    average_cpu_value = total_cpu_usage_value / number_of_readings_count
    print("Average CPU Load:", round(average_cpu_value), "%")
    if average_cpu_value < 50:
        print("Server Status: Normal")
    elif average_cpu_value < 80:
        print("Server Status: Warning")
    else:
        print("Server Status: Critical")

cpu_monitoring_data = [45, 60, 70, 85, 90]
classify_server_load(cpu_monitoring_data)
```

Problem Statement 5:Smart Classroom Resource Usage Monitor

```python
def monitor_classroom_energy(resource_usage_data):
    overused_list_container = []
    for resource_name_key in resource_usage_data:
        usage_hours_value = resource_usage_data[resource_name_key]
        if usage_hours_value > 8:
            overused_list_container.append(resource_name_key)
    if len(overused_list_container) > 0:
        print("Overused Resources:", ", ".join(overused_list_container))
        print("Energy Alert: Yes")
    else:
        print("Overused Resources: None")
        print("Energy Alert: No")
smart_class_usage_log = {
    "Projector": 6,
    "AC": 9,
    "Lights": 4
}
monitor_classroom_energy(smart_class_usage_log)
```

Problem Statement 6: Online Event Registration Capacity Controller

```python
def manage_event_registration(maximum_event_slots, total_registration_count):
    confirmed_count_tracker = 0
    waitlist_counter_value = 0
    for applicant_number in range(total_registration_count):
        if confirmed_count_tracker < maximum_event_slots:
            confirmed_count_tracker += 1
        else:
            waitlist_counter_value += 1
    print("Confirmed Registrations:", confirmed_count_tracker)
    print("Waitlisted Users:", waitlist_counter_value)
    if waitlist_counter_value > 0:
        print("Registration Status: Closed")
    else:
        print("Registration Status: Open")
event_capacity_limit = 100
registration_input_count = 105

manage_event_registration(event_capacity_limit, registration_input_count)
```
