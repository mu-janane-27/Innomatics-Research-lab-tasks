Problem Statement 1: Employee Performance Bonus Eligibility

```python
records = {
    "Ravi": 92,
    "Anita": 88,
    "Kiran": 92,
    "Suresh": 85
}

highest = max(records.values())
selected = []
for person in records:
    if records[person] == highest:
        selected.append(person)
print("Top Performers Eligible for Bonus:",", ".join(selected),"(Score:", highest, ")")
```

Problem Statement 2: Search Query Keyword Analysis

```python
text_data = "Buy mobile phone buy phone online"
text_data = text_data.lower()
word_list = text_data.split()
count_store = {}
for w in word_list:
    if w in count_store:
        count_store[w] += 1
    else:
        count_store[w] = 1
final_answer = {}
for w in count_store:
    if count_store[w] > 1:
        final_answer[w] = count_store[w]

print(final_answer)
```

Problem Statement 3: Sensor Data Validation

```python
sensors_readings = [3, 4, 7, 8, 10, 12, 5]
acceptable_data = []
for index in range(len(sensors_readings)):
    if sensors_readings[index] % 2 == 0:
        acceptable_data.append((index, sensors_readings[index]))
print("Valid Sensor Readings (Hour, Value):")
print(acceptable_data)
```

Problem Statement 4: Email Domain Usage Analysis

```python
mail_ids = [
    "ravi@gmail.com",
    "anita@yahoo.com",
    "kiran@gmail.com",
    "suresh@gmail.com",
    "meena@yahoo.com"
]

counting = {}

for item in mail_ids:
    domain_part = item[item.index("@") + 1:]

    if domain_part not in counting:
        counting[domain_part] = 1
    else:
        counting[domain_part] += 1

total_count = len(mail_ids)

print("Domain Usage Percentage:")

for key in counting:
    percentage_value = (counting[key] / total_count) * 100
    print(key + ":", str(int(percentage_value)) + "%")
```

Problem Statement 5: Sales Spike Detection

```python
daily_sales = [1200, 1500, 900, 2200, 1400, 3000]
sum = 0
for amount in daily_sales:
    sum += amount
average_sales = sum / len(daily_sales)
limit = average_sales * 1.30
for index in range(len(daily_sales)):
    if daily_sales[index] > limit:
        print("Day", index + 1, ":", daily_sales[index])
```

Problem Statement 6: Duplicate User ID Detection

```python
users = ["user1", "user2", "user1", "user3", "user1", "user3"]
printed = []
for u in users:
    if users.count(u) > 1 and u not in printed:
        print(u, "→", users.count(u), "times")
        printed.append(u)
```
