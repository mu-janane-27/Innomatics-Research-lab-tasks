Task 1: Replace Hardcoded Prompts (10%)\

```python
# Template
template_task1 = "Explain {topic} in simple terms for beginners"

prompt_task1 = PromptTemplate(
    input_variables=["topic"],
    template=template_task1
)

def explain_topic(topic):
    return prompt_task1.format(topic=topic)

# Output
print(explain_topic("AI"))
```

Task 2: Multi-Input Prompt System (15%)


```python
template_task2 = "Explain {topic} for {audience} in a {tone} tone"

prompt_task2 = PromptTemplate(
    input_variables=["topic", "audience", "tone"],
    template=template_task2
)

def generate_multi_prompt(topic, audience, tone):
    return prompt_task2.format(topic=topic, audience=audience, tone=tone)

# Test Cases
print(generate_multi_prompt("AI", "beginners", "friendly"))
print(generate_multi_prompt("Python", "kids", "fun"))
print(generate_multi_prompt("Deep Learning", "engineers", "technical"))
```

Task 3: Prompt Variations Engine (15%)

```python
# Teaching Template
teaching_template = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} clearly step by step"
)

# Interview Template
interview_template = PromptTemplate(
    input_variables=["topic"],
    template="Ask 3 questions about {topic}"
)

# Storytelling Template
story_template = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} as a story"
)

topic = "Machine Learning"

print(teaching_template.format(topic=topic))
print(interview_template.format(topic=topic))
print(story_template.format(topic=topic))
```

Task 4: ChatPromptTemplate System (15%)

```python
def get_chat_prompt(role, topic):
    system_roles = {
        "teacher": "You are a helpful teacher.",
        "interviewer": "You are an interviewer asking questions.",
        "motivator": "You motivate and inspire learners."
    }

    chat_template = ChatPromptTemplate.from_messages([
        ("system", system_roles[role]),
        ("user", "Talk about {topic}")
    ])

    return chat_template.format_messages(topic=topic)

# Example
messages = get_chat_prompt("teacher", "Neural Networks")
print(messages)
```

Task 5: Input Validation Layer (10%)


```python
def validate_inputs(audience, tone):
    valid_audience = ["beginner", "intermediate", "expert"]
    valid_tone = ["formal", "casual", "fun"]

    if audience not in valid_audience:
        audience = "beginner"

    if tone not in valid_tone:
        tone = "formal"

    return audience, tone

# Test
print(validate_inputs("kids", "fun"))   # invalid audience
```

Task 6: Prompt Generator App (15%)

```python
def generate_prompt(topic, audience, tone, style):

    # Validate inputs
    audience, tone = validate_inputs(audience, tone)

    styles = {
        "teaching": "Explain {topic} for {audience} in a {tone} teaching style",
        "interview": "Ask questions about {topic} for {audience} in a {tone} interview style",
        "storytelling": "Explain {topic} for {audience} in a {tone} storytelling style"
    }

    template = styles[style]

    prompt = PromptTemplate(
        input_variables=["topic", "audience", "tone"],
        template=template
    )

    return prompt.format(topic=topic, audience=audience, tone=tone)


# Example
print(generate_prompt("Neural Networks", "beginner", "fun", "storytelling"))
```
