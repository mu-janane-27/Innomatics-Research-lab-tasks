Problem 1: Smart Light Controller

```python
class SmartLightController:
    def __init__(self, device_label):
        self.device_label = device_label
        self.current_state = "OFF"

    def switch_on(self):
        self.current_state = "ON"

    def switch_off(self):
        self.current_state = "OFF"

    def show_status(self):
        print(self.device_label, "is", self.current_state)

bedroom_light = SmartLightController("Bedroom Light")
bedroom_light.switch_on()
bedroom_light.show_status()
```

Problem 2: Employee ID Card System

```python
class EmployeeIDCard:

    def __init__(self, employee_name, employee_code, employee_department):
        self.employee_name = employee_name
        self.employee_code = employee_code
        self.employee_department = employee_department

    def display_card_details(self):
        print("Employee ID Card")
        print("Name:", self.employee_name)
        print("ID:", self.employee_code)
        print("Department:", self.employee_department)

employee_object = EmployeeIDCard("Rahul", "EMP102", "IT")
employee_object.display_card_details()
```

Problem 3: Mobile Contact Record

```python
class MobileContact:

    def __init__(self, contact_person_name, contact_phone_number):
        self.contact_person_name = contact_person_name
        self.contact_phone_number = contact_phone_number

    def display_contact_details(self):
        print("Contact Saved")
        print("Name:", self.contact_person_name)
        print("Phone:", self.contact_phone_number)

contact_entry = MobileContact("Anita", "9876543210")
contact_entry.display_contact_details()
```

Problem 4: Product Price Tag Generator


```python
class ProductPriceTag:

    def __init__(self, product_label, product_price_value):
        self.product_label = product_label
        self.product_price_value = product_price_value

    def print_price_tag(self):
        print("Product:", self.product_label)
        print("Price: ₹" + str(self.product_price_value))

item_entry = ProductPriceTag("Headphones", 2499)
item_entry.print_price_tag()
```

Problem 5: Movie Rating Display System

```python
class MovieDetails:

    def __init__(self, movie_title_value, movie_rating_value):
        self.movie_title_value = movie_title_value
        self.movie_rating_value = movie_rating_value

    def display_movie_info(self):
        print("Movie:", self.movie_title_value)
        print("Rating:", self.movie_rating_value, "/ 5")

movie_object = MovieDetails("Inception", 4.8)
movie_object.display_movie_info()
```

Problem 6: Delivery Address Manager

```python
class DeliveryAddressManager:

    def __init__(self, customer_name_value, address_value):
        self.customer_name_value = customer_name_value
        self.address_value = address_value

    def display_delivery_details(self):
        print("Delivery Details")
        print("Customer:", self.customer_name_value)
        print("Address:", self.address_value)

delivery_record = DeliveryAddressManager("Suman", "Hyderabad")
delivery_record.display_delivery_details()
```
